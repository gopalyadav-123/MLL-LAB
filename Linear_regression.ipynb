```python
# Linear Regression
# y = b0 + b1*x

x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]

n = 5

sum_x = 0
sum_y = 0
sum_xy = 0
sum_x2 = 0

# Calculate required sums
for i in range(n):
    sum_x = sum_x + x[i]
    sum_y = sum_y + y[i]
    sum_xy = sum_xy + x[i] * y[i]
    sum_x2 = sum_x2 + x[i] * x[i]

# Calculate slope
b1 = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x * sum_x)

# Calculate intercept
b0 = (sum_y - b1 * sum_x) / n

print("Slope =", b1)
print("Intercept =", b0)

print("Regression Equation:")
print("y =", b0, "+", b1, "x")

# Prediction
x_new = 6
y_new = b0 + b1 * x_new

print("For x =", x_new)
print("Predicted y =", y_new)
```
